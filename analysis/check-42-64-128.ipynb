```python
import pandas as pd
import os


def generate_fixed_samples(input_file, output_folder='check_sample', num_samples=3, sample_size=100):
    """
    Sample data using fixed random seeds to ensure reproducibility.
    """
    if not os.path.exists(input_file):
        print(f"❌ Error: Source data file not found: {input_file}")
        return

    if not os.path.exists(output_folder):
        os.makedirs(output_folder)

    try:
        df = pd.read_csv(input_file)
        print("📊 Data loaded successfully. Preparing random sampling with fixed seeds...")
    except Exception as e:
        print(f"❌ Error reading the file: {e}")
        return

    # Use three different fixed seeds to obtain reproducible samples.
    seeds = [42, 64, 128]

    for i in range(num_samples):
        # Use the specified random seed.
        current_seed = seeds[i]
        sample = df.sample(n=sample_size, random_state=current_seed)

        file_name = f"sample_check_{i+1}_seed{current_seed}.csv"
        output_path = os.path.join(output_folder, file_name)

        sample.to_csv(output_path, index=False, encoding='utf-8-sig')
        print(f"   ✅ Sample file generated (seed {current_seed}): {output_path}")


if __name__ == "__main__":
    generate_fixed_samples('../dataset_tadf/all_data_with_smiles.csv')
    print("\n✨ Sampling complete! These three files are reproducible and can be reused for accuracy checks.")
```
